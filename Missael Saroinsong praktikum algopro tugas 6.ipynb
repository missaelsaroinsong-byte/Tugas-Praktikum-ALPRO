```python
# Variabel global
skor_nobita = 0

# Fungsi memperbarui skor
def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai

# Fungsi belajar
def belajar(jam_belajar):
    if jam_belajar < 1:
        print("Nobita hanya membuka buku sebentar")
        return 50
    elif jam_belajar < 3:
        print("Nobita belajar secukupnya")
        return 25
    else:
        print("Nobita belajar dengan rajin hari ini")
        return 10

# Fungsi tidur
def tidur(jam_tidur):
    if jam_tidur < 1:
        print("Nobita kurang tidur")
        return -10
    elif jam_tidur <= 2:
        print("Waktu tidur Nobita seimbang")
        return 20
    else:
        print("Nobita tidur terlalu lama")
        return -20

# Fungsi bermain
def main_bersama_doraemon(jumlah_gadget, waktu_main):
    if waktu_main <= 1 and jumlah_gadget <= 2:
        print("Nobita bermain secukupnya")
        return 15
    elif waktu_main <= 2 and jumlah_gadget <= 3:
        print("Nobita masih bisa mengatur waktu dengan baik")
        return 5
    else:
        print("Nobita kebanyakan bermain")
        return -30

# Fungsi menyontek
def menyontek():
    print("Nobita ketahuan menyontek")
    return -50

# Fungsi pengendali aktivitas
def jalankan_aktivitas(pilihan):
    if pilihan == 1:
        jam = float(input("Berapa jam Nobita belajar hari ini? "))
        nilai = belajar(jam)
        update_skor(nilai)

    elif pilihan == 2:
        jam = float(input("Berapa jam Nobita tidur siang hari ini? "))
        nilai = tidur(jam)
        update_skor(nilai)

    elif pilihan == 3:
        gadget = int(input("Berapa gadget yang digunakan Nobita? "))
        waktu = float(input("Berapa jam Nobita bermain gadget? "))
        nilai = main_bersama_doraemon(gadget, waktu)
        update_skor(nilai)

    elif pilihan == 4:
        nilai = menyontek()
        update_skor(nilai)

    else:
        print("Pilihan tidak valid!")

# Fungsi evaluasi akhir
def evaluasi_akhir(skor):
    if skor >= 80:
        return "Siswa Teladan"
    elif skor >= 50:
        return "Cukup Rajin"
    else:
        return "Pemalas"

# Menu utama
def menu_utama():
    global skor_nobita

    while True:
        print("\n=== PROGRAM PENINGKAT DISIPLIN NOBITA ===")
        print("1. Belajar")
        print("2. Tidur")
        print("3. Main bersama Doraemon")
        print("4. Menyontek saat ujian")

        pilihan = int(input("\nPilih aktivitas (1-4): "))

        if pilihan >= 1 and pilihan <= 4:
            jalankan_aktivitas(pilihan)
            print("Skor Nobita saat ini:", skor_nobita)

        else:
            print("Pilihan tidak valid, aktivitas hanya ada 4 pilihan")

        lanjut = input("\nApakah ingin melanjutkan? (y/n): ")

        if lanjut.lower() != "y":
            break

    print("\n=== HASIL EVALUASI NOBITA ===")
    print("Total skor:", skor_nobita)
    print("Evaluasi:", evaluasi_akhir(skor_nobita))

# Menjalankan program
menu_utama()
```
